```sql
%python
path = "/Volumes/fraud_kafka_project/bronze/fraud_watchlist/source/data"
```

```sql
%python
dbutils.fs.ls(path)
```

```sql
%python
df = spark.readStream.format("cloudFiles")\
                     .option("cloudFiles.format", "json")\
                     .option("cloudFiles.schemaLocation","/Volumes/fraud_kafka_project/bronze/fraud_watchlist/source/Schema_location")\
                     .option("inferSchema", "true")\
                     .load(path)
```

```sql
%python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```sql
%python
df = df.select(
    "*",
    col("_metadata.file_path").alias("file_path"),
    current_timestamp().alias("inserted_date")
)
```

```sql
%python
df.writeStream.format("delta")\
              .option("checkpointLocation", "/Volumes/fraud_kafka_project/bronze/fraud_watchlist/source/checkpoint")\
              .outputMode("append")\
              .trigger(availableNow=True)\
              .toTable("fraud_kafka_project.bronze.fraud_watchlist")
```
